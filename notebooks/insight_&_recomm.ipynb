## Insights (Drivers & Pain Points)
Objective: From your labeled and themed reviews, extract meaningful insights about what users like (drivers) and dislike (pain points) about the mobile banking apps.

identified_theme (e.g., 'Login Issues', 'User Interface', etc.)


```python
import pandas as pd

# Load CSV
df = pd.read_csv('../data/combined_analysis_results.csv')

# Normalize sentiment labels to lowercase
df['sentiment_label'] = df['sentiment_label'].str.lower()

# Group by theme and sentiment
theme_summary = df.groupby(['identified_theme', 'sentiment_label']).size().unstack(fill_value=0)

# Display top themes with most negative reviews
if 'negative' in theme_summary.columns:
    top_themes = theme_summary.sort_values(by='negative', ascending=False)
    print(top_themes.head(10))
else:
    print("No 'negative' sentiment entries found.")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(data=df, x='identified_theme', hue='sentiment_label')
plt.title('Sentiment Distribution by Bank')
plt.ylabel('Review Count')
plt.show()
```

```python
sns.histplot(data=df, x='sentiment_score', hue='sentiment_label', kde=True, bins=20)
plt.title("Sentiment Score Distribution per Bank")
plt.xlabel("Sentiment Score")
plt.show()
```

# Bank Recommendations

## 1. Commercial Bank of Ethiopia (CBE)
- **Recommendation:** Reliable choice with a strong physical presence, though the mobile app can be inconsistent.
- **Pros:**
  - Extensive branch network.
  - Essential banking functions available on the mobile app.
  - Generally reliable, but can be buggy.
- **Cons:**
  - Mobile app synchronization issues reported.
  - Mixed reviews on user-friendliness.
